### 1.1: Derivation of the Ordinary Least Squares Estimator with Calculus
*You should be able to work through this on your own anytime after lecture 2.*

🔗 In lecture, you learned the results of this derivation (the OLS estimator and intercept equations) and were introduced to the idea of "optimizing" parameters to minimize errors using two functions:

> 1) a function to "propose" new parameters \\
> 2) a function to "evaluate" how well the parameters work on data

Now we will apply those steps to derive an analytical result for OLS.

Let's say we have $n$ observations, each with some $x_i$ predictor and $y_i$ output. We would like to produce a set of predictions $\hat y$ of the form $\hat y_i = \beta_1 x_i + \beta_0$. (This is the function to "propose" parameters).

But we'd like them to be good predictions! So we want to derive $\beta_0$ and $\beta_1$ to *minimize error* between our true values $y_i$ and our predicted values $\hat y_i$. In fact, we will treat positive and negative errors (or residuals) equally and minimize the sum of squared errors, which is always a positive number. We can cleverly name this sum of squared errors $E$, to stand for "Error."

Since we'd like $\hat y_i$ to take the form $\beta_1 x_i + \beta_0$ (we're *assuming* a linear model), we can write the individual error for a the $i^\mathrm{th}$ datapoint as $(y_i - \beta_0 - \beta_1 x_i)$, which is just a stepping stone to writing the total sum of squared errors for an arbitrary model across all the data points as

$E(\beta_0, \beta_1) = \frac{1}{n} \sum_{i=1}^{n} (y_i - \beta_0 - \beta_1 x_i)^2$.

This error is a function of 2 variables: $\beta_0$ and $\beta_1$. By assuming a model, we have easily reframed the problem with calculus - we just need to search the parameter space to find where $E$ is minimized! The function is minimized where the partial derivatives are equal to 0 (if it can be minimized at all). In the case of this model, we can analytically solve the optimization problem we've created.


❓1.1.3
> You'll need a couple mathematical facts/tricks that will save some time:  \\
>> a) The sum of the $x$ or $y$ values is equal to $n$ times their mean, or $\sum_{i=1}^{n} y_i = n\bar y$.

>> b) $\left( \sum_{i=1}^{n} \left(x_i y_i \right) \right) - n\bar x \bar y = \left( \sum_{i=1}^{n} (x_i - \bar x) (y_i - \bar y) \right)$

>> c) $\sum_{i=1}^{n} \left(x_i^2 \right) - n \bar x^2 = \sum_{i=1}^{n} (x_i - \bar x)^2$


> *Keep in mind the question asks you to set both partial derivatives equal to 0 and solve. The facts above will be useful along the way.*

      

### ⭐ 1.2: Bonus - Linear Algebra Notation for Multiple Linear Regression

⭐ 1.2.1 In the derivation of ordinary least squares, you solved an optimization problem of 2 variables, using a system of two linear equations to find $\beta_0$ and $\beta_1$. Re-writing this problem using vectors and matrices (as below) can help us write compact equations for an arbitrary number of variables.

\begin{align}
  \hat Y &= \begin{pmatrix}  
             1 & x_1 \\
             1 & x_2 \\
             \vdots & \vdots \\
             1 & x_n
          \end{pmatrix}\begin{pmatrix}
                               \beta_0 \\
                               \beta_1
                              \end{pmatrix} \\~\\
         &= \begin{pmatrix}  
             1 & | \\
             \vdots & \vec x \\
             1 & |
          \end{pmatrix}\begin{pmatrix}
                               \beta_0 \\
                               \beta_1
                              \end{pmatrix} \\ ~\\
         &= X  \vec{\beta}
\end{align}

⭐ 1.2.2 When expanding to more independent variables, we simply have to add columns to our feature matrix.

⭐1.2.3
Above, we've defined an $X$ matrix and $\vec \beta$ vector that allow us to write the equation for residuals very simply as a function of the parameter vector, $\vec \beta$:
\begin{align}
 \vec e(\vec \beta) &= Y - \hat Y \\
               &= Y - X \vec \beta
\end{align}

Keep in mind that this error $\vec e(\vec \beta)$ is a *vector* of residuals in $\mathbb{R}^n$ and not the single sum scalar we had in the previous problem, $E$. This lets us write the total sum of squared residuals, $E$, using a dot product: $E = \vec e \cdot \vec e = \vec e^T \vec e$.

*Big Hint: No matter how many dimensions we're working with, matrices help us simplify our notation, hiding the high dimensions we're working in so that we don't have to adjust how we think about error vector with every additional dimension.*

### 2.5: Assessing the Model
### Visualizing Error
To do linear regression, we assume that there is a linear relationship between $x$ and $y$, and we assume that the errors in $y$ are normally and equally distributed at each $x$ value. One of the easiest ways to visualize whether these assumptions hold in our model is to make a residuals plot, which displays the errors of the model (the residuals, $y - \hat y$) against the x-values themselves.

We can also use a Q-Q plot to assess more globally if the residuals are normally distributed. Unlike a residuals scatterplot, you won't be able to visualize the residuals as a function of $x$.
